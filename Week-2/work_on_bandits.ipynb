# Bandit Problem

```python
# imports
from bandits import Bandit
import random
# Include your imports here, if any are used. 
import numpy as np
import matplotlib.pyplot as plt
```

A list of ten bandit objects initialized in the list...

```python
bandits = [Bandit(random.random()*4-2) for _ in range(10)]
```

```python
bandits[0].pullLever()
```

## Greedy algorithm Implementation

```python
def run_greedy():
    # TODO: Implement the greedy algorithm here
    # Return the reward from the bandits in a list
    rewards = []
    Q = np.zeros(len(bandits))
    N = np.zeros(len(bandits))

    for _ in range(1000):
        a = np.argmax(Q)
        r = bandits[a].pullLever()
        N[a] += 1
        Q[a] += (r - Q[a]) / N[a]
        rewards.append(r)

    return rewards

    
```

Plot the cumulative average of rewards as the number of iterations increases.

```python
greedy_rewards = run_greedy()
cumulative_average_greedy = np.cumsum(greedy_rewards) / (np.arange(len(greedy_rewards)) + 1)

plt.figure(figsize=(10, 6))
plt.plot(cumulative_average_greedy, label='Greedy')
plt.xlabel('Iterations')
plt.ylabel('Cumulative Average Reward')
plt.title('Greedy Algorithm Performance')
plt.legend()
plt.show()
```

## $\epsilon$-greedy Algorithm

```python
def run_epsilon_greedy(epsilon):
    # TODO: Implement the epsilon greedy algorithm here
    # Return the reward from the bandits in a list
    rewards = []
    Q = np.zeros(len(bandits))
    N = np.zeros(len(bandits))
    
    for _ in range(1000):
        p = np.random.random()
        if p < epsilon:
            a = np.random.randint(len(bandits))
        else:
            a = np.argmax(Q)
            
        r = bandits[a].pullLever()
        N[a] += 1
        Q[a] += (r - Q[a]) / N[a]
        rewards.append(r)
    
    return rewards
```

Plot the cumulative average of rewards as the number of iterations increases but for various values of $\epsilon$.

```python
epsilons = [0.01, 0.1, 0.5]

plt.figure(figsize=(10, 6))
for eps in epsilons:
    rewards = run_epsilon_greedy(eps)
    cumulative_average = np.cumsum(rewards) / (np.arange(len(rewards)) + 1)
    plt.plot(cumulative_average, label=f'epsilon={eps}')

plt.xlabel('Iterations')
plt.ylabel('Cumulative Average Reward')
plt.title('Epsilon-Greedy Performance')
plt.legend()
plt.show()
```

## Finding the optimal $\epsilon$

Run the $\epsilon$-greedy algorithm for 1000 iterations and find the optimal $\epsilon$ value by plotting the cumulative average of rewards for various values of $\epsilon$

```python
epsilons = [0, 0.01, 0.05, 0.1, 0.2, 0.5, 1.0]

plt.figure(figsize=(10, 6))
for eps in epsilons:
    rewards = run_epsilon_greedy(eps)
    cumulative_average = np.cumsum(rewards) / (np.arange(len(rewards)) + 1)
    plt.plot(cumulative_average, label=f'epsilon={eps}')

plt.xlabel('Iterations')
plt.ylabel('Cumulative Average Reward')
plt.title('Performance Comparison for Different Epsilons')
plt.legend()
plt.show()
```

## Optimistic Initial Values

```python
def run_optimistic_greedy():
    # TODO: Implement the optimistic greedy algorithm here
    rewards = []
    initial_value = 10.0
    Q = np.full(len(bandits), initial_value)
    N = np.zeros(len(bandits))
    
    for _ in range(1000):
        a = np.argmax(Q)
        
        r = bandits[a].pullLever()
        N[a] += 1
        Q[a] += (r - Q[a]) / N[a]
        rewards.append(r)
    
    # Return the reward from the bandits in a list
    return rewards
    
```

Plot the cumulative average of rewards as the number of iterations increases for an optimistic greedy of $Q_1 = 10$ and a non-optimistic $\epsilon = 0.1$

```python
# Run Optimistic
opt_rewards = run_optimistic_greedy()
opt_cumulative = np.cumsum(opt_rewards) / (np.arange(len(opt_rewards)) + 1)

# Run Epsilon Greedy (using 0.1 as comparison)
eps_rewards = run_epsilon_greedy(0.1)
eps_cumulative = np.cumsum(eps_rewards) / (np.arange(len(eps_rewards)) + 1)

plt.figure(figsize=(10, 6))
plt.plot(opt_cumulative, label='Optimistic Greedy (Q1=5, eps=0)')
plt.plot(eps_cumulative, label='Epsilon-Greedy (eps=0.1)')
plt.xlabel('Iterations')
plt.ylabel('Cumulative Average Reward')
plt.title('Optimistic Initial Values vs Epsilon-Greedy')
plt.legend()
plt.show()
```
